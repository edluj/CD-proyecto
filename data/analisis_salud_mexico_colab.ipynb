```python
from pathlib import Path
import json
import importlib.util
import pandas as pd
from IPython.display import Image, Markdown, display

candidate_dirs = [
    Path('/content/drive/MyDrive/avance de proyecto'),
    Path('/content/drive/MyDrive/avance_de_proyecto'),
]

BASE_DIR = None
for candidate in candidate_dirs:
    if candidate.exists():
        BASE_DIR = candidate
        break

if BASE_DIR is None:
    raise FileNotFoundError(
        "No se encontro la carpeta 'avance de proyecto' en Google Drive. "
        "Subela a Mi unidad y vuelve a ejecutar."
    )

SCRIPT_PATH = BASE_DIR / 'analisis_salud_mexico.py'
OUTPUT_DIR = BASE_DIR / 'analisis_salud_mexico'

print('Carpeta base:', BASE_DIR)
print('Script:', SCRIPT_PATH)
print('Salida:', OUTPUT_DIR)
print('\nArchivos detectados:')
for p in sorted(BASE_DIR.iterdir()):
    print('-', p.name)
```

```python
if not SCRIPT_PATH.exists():
    raise FileNotFoundError(
        f"No se encontro el script {SCRIPT_PATH}. "
        "Copia tambien 'analisis_salud_mexico.py' dentro de la carpeta de Drive."
    )

spec = importlib.util.spec_from_file_location('analisis_salud_mexico_mod', SCRIPT_PATH)
mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(mod)

mod.main()
```

## Revision de valores faltantes

En esta seccion se revisa la presencia de datos faltantes dentro del archivo `salud_mexico_sintetica_75000.csv` usando `pandas` y `missingno`.


```python
import pandas as pd
import numpy as np
import missingno as msno
import matplotlib.pyplot as plt

df = pd.read_csv(OUTPUT_DIR / 'salud_mexico_sintetica_75000.csv', encoding='latin1')
df_imput03 = df.copy()

df_imput03.isnull().sum() / df_imput03.shape[0]
```
